```python
from collections import deque

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

root = Node('A')
root.left = Node('B')
root.right = Node('C')
root.left.left = Node('D')
root.left.right = Node('E')
root.right.left = Node('F')
root.right.right = Node('G')

queue = deque()
queue.append(root)

while queue:
    current = queue.popleft()
    print(current.data, end=" ")

    if current.left:
        queue.append(current.left)

    if current.right:
        queue.append(current.right)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def dfs(root):
    if root is None:
        return

    print(root.data, end=" ")
    dfs(root.left)
    dfs(root.right)

root = Node('A')
root.left = Node('B')
root.right = Node('C')
root.left.left = Node('D')
root.left.right = Node('E')
root.right.left = Node('F')
root.right.right = Node('G')

dfs(root)
```

```python
from collections import deque

def solve_grid_search(rows, cols, start, target, method='BFS'):
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    frontier = deque([start])
    visited = set([start])
    visited_count = 0

    parent = {start: None}

    while frontier:
        current = frontier.popleft() if method == 'BFS' else frontier.pop()
        visited_count += 1

        if current == target:
            path = []
            curr = target
            while curr is not None:
                path.append(curr)
                curr = parent[curr]
            return path[::-1], visited_count

        r, c = current
        for dr, dc in moves:
            nr, nc = r + dr, c + dc
            neighbor = (nr, nc)

            if 0 <= nr < rows and 0 <= nc < cols and neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = current
                frontier.append(neighbor)

    return None, visited_count

GRID_ROWS, GRID_COLS = 3, 3
START = (0, 0)
TARGET = (2, 2)

bfs_path, bfs_visited = solve_grid_search(GRID_ROWS, GRID_COLS, START, TARGET, method='BFS')
dfs_path, dfs_visited = solve_grid_search(GRID_ROWS, GRID_COLS, START, TARGET, method='DFS')

print(f"BFS Path: {bfs_path} | Total Visited States: {bfs_visited}")
print(f"DFS Path: {dfs_path} | Total Visited States: {dfs_visited}")
```
